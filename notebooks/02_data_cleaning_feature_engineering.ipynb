```python
import pandas as pd
import numpy as np

df = pd.read_csv("../dataset/GiveMeSomeCredit-training.csv")
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

df_clean = df.copy()
```

```python
median_age = df_clean['age'].median()
df_clean.loc[df_clean['age'] == 0 , 'age'] = median_age
```

```python
df_clean['age'].min()
```

```python
df_clean['MonthlyIncome_missing'] = df_clean['MonthlyIncome'].isnull().astype(int)
```

```python
median_income = df_clean['MonthlyIncome'].median()
df_clean['MonthlyIncome'] = df_clean['MonthlyIncome'].fillna(median_income)
```

```python
df_clean['MonthlyIncome'].isnull().sum()  # should be 0 now
```

```python
df_clean['MonthlyIncome_missing'].sum()   # should be ~29,731 (matches what we found earlier)
```

```python
df_clean['MonthlyIncome'].median()        # sanity check the value used
```

```python
df_clean['DebtRatio'].quantile([0.95, 0.99, 0.999])
```

```python
df_clean['DebtRatio'] = df_clean['DebtRatio'].clip(upper=3)
```

```python
df_clean['DebtRatio'].describe()
```

```python
df_clean['RevolvingUtilizationOfUnsecuredLines'] = df_clean['RevolvingUtilizationOfUnsecuredLines'].clip(upper=10)
```

```python
df_clean['RevolvingUtilizationOfUnsecuredLines'].describe()
```

```python
df_clean['has_sentinel_delinquency'] = (
    (df_clean['NumberOfTime30-59DaysPastDueNotWorse'].isin([96, 98])) |
    (df_clean['NumberOfTime60-89DaysPastDueNotWorse'].isin([96, 98])) |
    (df_clean['NumberOfTimes90DaysLate'].isin([96, 98]))
).astype(int)
```

```python
cols_days_late = [
    'NumberOfTime30-59DaysPastDueNotWorse',
    'NumberOfTime60-89DaysPastDueNotWorse',
    'NumberOfTimes90DaysLate'
]

for col in cols_days_late:
    df_clean[col] = df_clean[col].replace([96, 98], 0)
```

```python
df_clean[cols_days_late].describe()
df_clean['has_sentinel_delinquency'].sum()  # should be 269
```

```python
df_clean['NumberOfDependents'] = df_clean['NumberOfDependents'].fillna(0)
```

```python
df_clean['NumberOfDependents'].isnull().sum()  # should be 0
```

```python
df_clean['NumberRealEstateLoansOrLines'] = df_clean['NumberRealEstateLoansOrLines'].clip(upper=20)
```
